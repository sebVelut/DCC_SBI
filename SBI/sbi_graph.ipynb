# Raincloud of the accuracy

```python
from matplotlib import rcParams

def raincloud_graph(results,title,xlabel,xlim=None, save=False, savepath=None, legend=False):
    rcParams['font.weight'] = 'bold'

    x = "prior"
    y = "score"
    hue = "classification mode"

    boxplot_lw = 3.0
    boxplot_props = {'linewidth': boxplot_lw}

    temp = results.copy()
    # temp.insert(1,"procedure",np.repeat(["DG","DA","SS","DG","DA","SS"],12*1),True)

    n_pi = 3
    palette = dict(zip(["Sim2Emp","Sim2Sim"]*3, [sns.color_palette('Set2', 7)[0],sns.color_palette('Set2', 7)[2]]*3))
    fig, ax = plt.subplots(1, 1)
    fig.set_figheight(6)
    fig.set_figwidth(8)
    pt.RainCloud(
        data=results, y=y, x=x,
        hue=hue,
        bw='scott',
        width_viol=0, width_box=0.8, point_size=8,
        dodge=True, orient='h',
        linewidth=0, box_linewidth=boxplot_lw,
        box_whiskerprops=boxplot_props,
        box_medianprops=boxplot_props,
        alpha=0.7, palette=palette,
        box_showfliers=False,
        fontsize=25,
        ax=ax, pointplot=True,
        point_linestyles="none",
        point_markers='D',)



    # ax.set_yticks([])

    for i in range(n_pi):
        ax.axhline(y=i + 0.5, xmin=0, xmax=1.02, color='black' , linestyle=':', alpha=0.3)
    if xlim is not None:
        ax.set_xlim(left=0,right=xlim)
    # GREEN_patch = mpatches.Patch(color=sns.color_palette('Set2', 5)[0], label='Domain Generalization')
    # PTGREEN_patch = mpatches.Patch(color=sns.color_palette('Set2', 5)[1], label='Domain Adaptation')
    # SS_patch = mpatches.Patch(color=sns.color_palette('Set2', 5)[2], label='Single Subject')
    # SS_patch = mpatches.Patch(color=sns.color_palette('Set2', 5)[3], label='Single Subject')
    # SS_patch = mpatches.Patch(color=sns.color_palette('Set2', 5)[4], label='Single Subject')

    ax.set_xlabel(xlabel,fontsize=25, fontweight='bold')
    ax.set_ylabel("",fontsize=25, fontweight='bold')
    ax.set_title(title,fontsize=30, fontweight='bold')
    # ax.legend(['Domain Generalization', 'Domain Adaptation', 'Single Subject'],
    #         handles=[DG_patch,DA_patch,SS_patch],
    #             loc=(0.05, 1.01), ncols=n_pi)
    
    handles, labels = ax.get_legend_handles_labels()
    _ = plt.legend(handles[0:len(labels) // 4], labels[0:len(labels) // 4], \
                       bbox_to_anchor=(1.02, 1), loc=2, borderaxespad=0., \
                       title="", fontsize=20)  # , title_fontsize = 25)
    

    if not legend:
        ax.get_legend().remove()
    
    if save:
        plt.savefig(savepath,bbox_inches='tight',dpi=300)

```

```python
raincloud_graph(results_df[results_df["prior"]=="within-pp"],"","Accuracy",1.02, save=True, 
                savepath='D:/s.velut/Documents/These/DCC/SBI/results/overall_sbi/graphes/results_within.pdf',legend=True)
```
